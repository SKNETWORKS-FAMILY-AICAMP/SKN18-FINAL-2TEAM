# Data Cleaning
- 대상파일: "api_data_XXX.csv"
- url: 중복되는 값만 제거하는 것으로 처리
- title: url 중복제거 후 title이 중복된다면 제거(거의 abstract, step_content, referecnce …등 데이터가 완전히 일치함)
- abstract: “””” 등 여러개가 있는 경우 “ 한개로 변환 => replace이용, 띄어쓰기 2칸이상 들여쓰기 제거(예: “””summary: ⇒ “summary: ), \n, \t 등 특수 문자 제거(내용에 직접 기입된 것), enter가 두 번 입력된것(python print시 “\n\n”로 표시)는 enter 한 번 입력된 것으로 변경(replace(“\n\n, \n”)이용)
- step_content: 띄어쓰기가 2번이상 연속으로 된경우 띄어쓰기 한칸으로 처리, \n, \t 등 특수문자 제거(내용에 직접 기입된 것), <no data>포함된 데이터에서 다른 데이터가 있다면 <no data>제거 => <no data>를 일괄제거하는 것으로 해결가능(데이터가 없다면 빈칸), ‘\u003e’, ‘\u003c’와 같은 ‘\uXXX ‘형식의 키워드를 제거해야함
- reference: <no data>가 대부분 => <no data> 무조건 삭제, 따옴표 제거시 비어있다면 <no data>와 동일하게 처리됨(예: “ , “”, “” “”” “”) => 영어 알파벳 및 숫자가 포함되지 않는 경우에 해당
- guidelines: 상기의 step_content와 비슷함, 따옴표 제거 필요, NA(no data와 비슷함) => 제거
- materials: NA => 제거, <no data>나 length가 0인 경우 빈 데이터 처리(null 처리), 따옴표 연속2개이상인 경우 1개로 수정
- 공통사항: <no data>제거, 따옴표 여러개 있는 경우 하나로 수정, \n \t \uXXX 등 특수문자 제거, NA제거, 띄어쓰기 2칸이상 1칸으로 처리, 데이터 없음 => null로처리

# 대상 파일 지정

```python
# jupyter에서의 전처리는 하나의 파일로 진행 => 전처리 통합본은 py파일로 만들며, 모든 대상 파일로 함
import pandas as pd

df = pd.read_csv("api_data_DNA.csv")
df
```

# URL 중복 제거

```python
# url 중복확인
len(df["url"].unique())
```

```python
# url 중복제거 후 테이블
df = df.drop_duplicates(subset="url", keep="first").reset_index()
df = df.rename(columns={'index': 'protocol_id'})
df
```

# TITLE 중복 제거

```python
# title 중복 확인
len(df["title"].unique())
```

```python
# title 중복제거 후 테이블
df = df.drop_duplicates(subset="title", keep="first").reset_index()
df = df.drop(columns=['index'])
df
```

# 각 요소별 text 전처리

- “””” 등 여러개가 있는 경우 “ 한개로 변환

```python
df = df.replace(r'"+', '"', regex=True)
df = df.replace(r'" ', '', regex=True)
```

```python
# 전처리 후 확인
df
```

```python
# 띄어쓰기 2칸이상 들여쓰기 제거
df = df.replace(r'\s{2,}', ' ', regex=True)
df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)
df
```

- 특수 문자 제거(내용에 직접 기입된 것)

```python
df = df.replace(r"\t", "", regex=True)
df
```

- enter가 두 번 입력된것

```python
df = df.replace(r'\n+', '\n', regex=True)
df
```

- 문자열 "\<no data\>" 제거

```python
df = df.replace("<no data>", "")
df
```

-  ‘\u003e’, ‘\u003c’와 같은 ‘\uXXX ‘형식의 키워드를 제거

```python
df = df.replace(r"\\u[0-9a-fA-F]{4}", "", regex=True)
df
```

- “ , “”, “” “”” “”와 같이 따옴표외 문자 및 숫자가 없는 경우 => 빈 데이터 처리

```python
# 따옴표만 있거나 공백만 있는 경우 NaN으로 변환
df = df.replace(r'^\s*["“”\'’]*\s*$', "", regex=True)
df
```

- "NA"(no data와 비슷함) => 제거

```python
df = df.replace("NA", "")
df
```

# 데이터 값이 없는 경우(띄어쓰기만 있는 것도 포함)
- RDB 및 VectorDB 도입을 고려해야함
- 데이터 값이 없는 경우 null처리 => vectorDB에서 null 임베딩 오류 발생 => 권장하지 않음
- 데이터 값이 없는 경우 빈 문자열 처리 => 의미 없는 벡터 생성 => 결과 품질 저하
- 데이터 값이 없는 경우 "\<no data\>" 삽입(Placeholder) => \<no data\> 임베딩은 의미 없는 벡터 => semantic search에서 다른 쿼리와 거의 관계 없는 결과로 노이즈 발생 => 여러 문서가 \<no data\>이면 벡터들이 서로 비슷하게 만들어져 검색 결과가 섞임

- 결론: 데이터 값이 없는 경우 빈 문자열("")로 처리 => VectorDB 구축시 이를 어떻게 처리할지 고려 필요

```python
# NaN을 빈 문자열로 변경
df = df.fillna("")

# 공백만 있는 문자열을 빈 문자열로 변경
df = df.applymap(lambda x: "" if isinstance(x, str) and x.strip() == "" else x)
df
```

# 전처리 결과를 확인할 수 있게 csv파일로 생성

```python
df.to_csv(f"cleaned_data_DNA.csv", index=False)
```
