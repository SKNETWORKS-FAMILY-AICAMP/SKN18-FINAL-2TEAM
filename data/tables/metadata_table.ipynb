# 대상 파일: api_data_Cell.csv

```python
import pandas as pd

df = pd.read_csv("api_data_Cell.csv")
df
```

# dataframe의 좌측 숫자를 "protocol_id"로 설정

```python
df = df.reset_index().rename(columns={'index': 'protocol_sid'})
df
```

# 테이블1: t_protocol_metadata
- columns: protocol_id(PK), url, title

```python
table_metadata = df[["protocol_sid", "url", "title"]]
table_metadata
```

```python
# t_protocol_metadata 테이블1에 대한 csv파일생성
table_metadata.to_csv("t_protocol_metadata_Cell.csv", index=False)
```

# 테이블2: t_protocol_materials
- columns: materials_sid(PK), protocol_id(FK => t_protocol_metadata), materials

```python
table_materials = df[["protocol_sid", "materials"]]
table_materials
```

```python
# materials_sid(PK) 설정
table_materials = table_materials.reset_index().rename(columns={'index': 'materials_sid'})
table_materials
```

```python
# t_protocol_materials 테이블2에 대한 csv파일생성
table_materials.to_csv("t_protocol_materials_Cell.csv", index=False)
```

# 테이블3: t_protocol_references
- columns: references_sid(PK), protocol_id(FK), reference

```python
table_references = df[["protocol_sid", "reference"]]
table_references
```

```python
# materials_sid(PK) 설정
table_references = table_references.reset_index().rename(columns={'index': 'reference_sid'})
table_references
```

```python
# t_protocol_reference 테이블3에 대한 csv파일생성
table_references.to_csv("t_protocol_references_Cell.csv", index=False)
```
