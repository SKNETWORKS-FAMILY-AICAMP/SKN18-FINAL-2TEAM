```python
# 경로 설정 (프로젝트 루트를 sys.path에 추가)
import sys
import os
from pathlib import Path

# 현재 노트북 위치에서 프로젝트 루트까지 올라가기
project_root = Path.cwd().parent.parent  # graph/test -> graph -> project_root
sys.path.insert(0, str(project_root))

print(f"프로젝트 루트 경로: {project_root}")
print(f"현재 작업 디렉토리: {Path.cwd()}")

# 필요한 라이브러리 import
from IPython.display import Image, display
from graph.compile import create_workflow

# 워크플로우 생성
app = create_workflow()
print("✅ Medical RAG 워크플로우 생성 완료!")

# 워크플로우 그래프 시각화
try:
    # Mermaid PNG 방식 (권장)
    graph_image = app.get_graph().draw_mermaid_png()
    display(Image(graph_image))
    print("✅ 워크플로우 그래프 표시 완료")
except Exception as e:
    print(f"❌ Mermaid 그래프 표시 오류: {e}")
    
    # 대안 1: ASCII 그래프
    try:
        print("\n📊 ASCII 그래프로 대체:")
        print(app.get_graph().draw_ascii())
    except Exception as ascii_error:
        print(f"❌ ASCII 그래프도 실패: {ascii_error}")
        
        # 대안 2: 그래프 구조 정보만 출력
        try:
            print("\n📋 그래프 노드 정보:")
            graph = app.get_graph()
            print(f"노드 수: {len(graph.nodes)}")
            print(f"노드 목록: {list(graph.nodes.keys())}")
        except Exception as info_error:
            print(f"❌ 그래프 정보 출력도 실패: {info_error}")
            
```
